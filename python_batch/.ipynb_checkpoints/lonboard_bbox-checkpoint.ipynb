```python
import duckdb
from lonboard import viz
from gpq.utils import apply_s3_creds, install_load_extensions

install_load_extensions()
apply_s3_creds()
```

```python


# little cottonwood canyon
bbox = (-111.700745,40.541574,-111.584873,40.598978)
overture_version = "2025-03-19.1"



result = duckdb.sql(f"""SELECT subtype, height, roof_material, facade_material, facade_color, roof_shape, roof_orientation, geometry, bbox FROM read_parquet('s3://overturemaps-us-west-2/release/{overture_version}/theme=buildings/type=building/*.parquet')
WHERE
bbox.xmin BETWEEN {bbox[0]} AND {bbox[2]} AND
bbox.ymin BETWEEN {bbox[1]} AND {bbox[3]} """)


```
